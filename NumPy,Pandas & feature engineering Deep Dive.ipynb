```python
# Merge:
#Types of merges
# inner
# left
# right
# outer

df_1 = pd.DataFrame(data={
    "Name":["Monika","Rahul","Priya"],
    "Age" :[23,24,45],
    "Salary":[30000,35000,25000]
},index=[1, 2, 3])
df_1.index.name = "Emp_id"

df_2 = pd.DataFrame(data={
    "Name":["Raj","Rohan"],
    "Age": [34,65],
    "Salary": [58000,760000]
},index= [1,2])
df_2.index.name= "Emp_id"
```

```python
display(df_1)
display(df_2)
```

```python
#Inner:Only matching records.
merged_inner = pd.merge(
    df_1,
    df_2,
    on="Emp_id",
    how="inner"
)
display(merged_inner)
```

```python
# Left :Keep everything from left DataFrame.
merged_left = pd.merge(
    df_1,
    df_2,
    on="Emp_id",
    how="left"
)
display(merged_left)
```

```python
# Right :Keep everything from right DataFrame.
merged_right = pd.merge(
    df_1,
    df_2,
    on="Emp_id",
    how="right"
)
display(merged_right)
```

```python
#Outer:Keep everything from both.
merged_outer = pd.merge(
    df_1,
    df_2,
    on="Emp_id",
    how="outer"
)
display(merged_outer)
```

## Feature engineering:
Feature engineering is the process of converting raw data into useful input features that a machine-learning model can learn from effectively.

For example, suppose we have:

Date of Birth: 15-05-1995

A model generally shouldn't directly use this raw date.

We can create:

Age = 31

### Encoding

```python
#Label Encoding:
#say suppose we have gender :
# Gender
# ------
# Male
# Female
# Female
# Male

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

# will results into :
# Male   → 1
# Female → 0

# but never apply Label encoding on ordering.like:
# City:
# Pune
# Mumbai
# Delhi

# which will get converted into:
# Pune   → 0
# Mumbai → 1
# Delhi  → 2
# which means: Delhi > Mumbai > Pune

# which has no meaningful relationship.

# Therefore, LabelEncoder is generally not the right choice for nominal input features.
```

```python
#Ordinal Encoding:

from sklearn.preprocessing import OrdinalEncoder
# High School → 1
# Bachelor     → 2
# Master       → 3
# PhD          → 4

# This is appropriate because the ordering contains information.
```

```python
#One-Hot Encoding
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore"
)

# One-hot encoding creates:

# Pune | Mumbai | Delhi
# ---------------------
# 1    | 0      | 0
# 0    | 1      | 0
# 0    | 0      | 1

# Why handle_unknown="ignore"?:

# Suppose during training we saw:

# Pune
# Mumbai
# Delhi

# but during production we receive:

# Bangalore

# Without proper handling, the encoder can raise an error.
# handle_unknown="ignore"

# the unseen category is handled safely.

# This is an important production interview point.
```

### Mixed Numerical + Categorical Data

```python
# Suppose:

# Age              → numerical
# Salary           → numerical
# Experience       → numerical
# City             → categorical
# Education        → categorical

# We can use:

from sklearn.compose import ColumnTransformer

numeric_features = [
    "Age",
    "Salary",
    "Experience"
]

categorical_features = [
    "City",
    "Education"
]
# numeric pipeline
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical pipeline:

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore"
    ))
])

# Combine:

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

# Then model:

model_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

# This is the kind of structure you should be comfortable explaining in an ML interview.
```
